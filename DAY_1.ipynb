```python
name = input("Enter student name: ")
age = int(input("Enter student age: "))

marks1 = float(input("Enter marks in Subject 1: "))
marks2 = float(input("Enter marks in Subject 2: "))
marks3 = float(input("Enter marks in Subject 3: "))

total = marks1 + marks2 + marks3
average = total / 3
percentage = (total/300)*100

if marks1 >= 40 and marks2 >= 40 and marks3 >= 40 and average >= 60:
    merit = "Eligible for Merit"
else:
    merit = "Not Eligible for Merit"

print("\n--- Student Details ---")
print("Name:", name)
print("Age:", age)
print("Subject 1 Marks:", marks1)
print("Subject 2 Marks:", marks2)
print("Subject 3 Marks:", marks3)

print("\n--- Result ---")
print("Total Marks:", total)
print("Average:", average)
print("Percentage:", percentage)
print("Merit Eligibility:", merit)

print("\n--- Data Types ---")
print("Data type of name:", type(name))
print("Data type of age:", type(age))
print("Data type of marks:", type(marks1))
print("Data type of average:", type(average))
```

```python
#acticty : exploring list funciton using a fruit list

fruits=["Apple","Banana","Mango","Orange"]
print(fruits)
print(len(fruits))

fruits.append("Grapes")
more_fruits = ["Pineapple","Cherry"]
fruits.extend(more_fruits)
print("After extend:",fruits)

print("First fruit:",fruits[0])
print("Last fruit:",fruits[-1])
print(fruits[2:5])

fruits[1] = "Strawbeery"
print("After updating index 1:", fruits)
fruits.remove("Mango")
print("After remove Mango:",fruits)

removed_fruit = fruits.pop(2)
print(removed_fruit)
print("After pop:", fruits)


fruits.sort()
print("After sort:", fruits)
fruits.reverse()
print("After reverse:",fruits)


fruits.clear()
print(fruits)

```
